```python
# closures
A function object that remembers
values from its outer function
even after the outer function has finished execution.





A closure happens when:

1. A function is defined inside another function
2. Inner function uses outer function variables
3. Outer function returns inner function
4. Inner function still remembers outer variables
```

```python
def multiplier(x):

    def multiply(y):
        return x * y

    return multiply


double = multiplier(2)

triple = multiplier(3)

print(double(5))

print(triple(5))
```

```python
def outer():

    x = 10

    def inner():
        print(x)

    return inner


func = outer()

print(func.__closure__)
print(func.__closure__[0].cell_contents)
```

```python
def rate_limiter(limit):

    calls = 0

    def api():

        nonlocal calls

        if calls >= limit:
            print("Limit Exceeded")

        else:
            calls += 1
            print(f"API Call {calls}")

    return api


user_api = rate_limiter(3)

user_api()
user_api()
user_api()
user_api()
```

```python
def logger(func):

    def wrapper():

        print("Before Function")

        func()

        print("After Function")

    return wrapper


@logger
def hello():
    print("Hello")


hello()
```
