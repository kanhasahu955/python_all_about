```python
# method overloading
# Same method name
# but different parameters.



# class Test:

#     def add(self, a, b):
#         print(a+b)

#     def add(self, a, b, c):
#         print(a+b+c)

# obj = Test()

# obj.add(1,2)



# class Math:

#     def add(self, a, b, c=0):
#         print(a+b+c)

# obj = Math()

# obj.add(1,2)

# obj.add(1,2,3)


class Math:

    def add(self, *args):

        total = 0

        for i in args:
            total += i

        print(total)

obj = Math()

obj.add(1,2)

obj.add(1,2,3)

obj.add(1,2,3,4)
```

```python
| Overloading                  | Overriding                   |
| ---------------------------- | ---------------------------- |
| Same method different params | Child modifies parent method |
| Compile-time                 | Runtime                      |
| Partial support in Python    | Fully supported              |
```
