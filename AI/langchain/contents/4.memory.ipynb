```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-5.4")

prompts = [
    {
        "role": "user",
        "content": "hi i am ashok sahu",
    },
    {
        "role": "ai",
        "content": "Hi Ashok Sahu! Nice to meet you. How can I help you today?",
    },
    {"role": "user", "content": "what is my name?"},
]

result = llm.invoke(prompts).content
print(result)
```

<h1>History track</h1>

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model='gpt-5.4')

history:list = []

while True:
    query = input("user: ",)

    words = [chunk.strip(".,!?;:'\"") for chunk in query.lower().split()]
    if any([w in words for w in ["bye", "exit", "quit"]]):
        print("good bye")
        break

    history.append({"role": "user", "content": query})
    print("User: ",query)

    result = llm.invoke(history)
    history.append({"role":"ai","content":result.content})
    print("AI: ",result.content,"\n")
```
