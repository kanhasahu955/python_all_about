<h1>Dynamic Prompt Template</h1>

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model='gpt-5.4')
prompts = [
    {"role":"system","content":"you are a translator and your translating the content to {language}"},
    {"role":"user","content":"{query}"}
]

prompt_template = ChatPromptTemplate.from_messages(prompts)
# final_prompt = prompt_template.format(language='odia',query='How are you')
final_prompt = prompt_template.format(language='odia',query='How are you')
final_prompt = prompt_template.format(language='japan',query='How are you')
final_prompt = prompt_template.format(language='English',query='How are you')
final_prompt = prompt_template.format(language='Hindenglish',query='How are you')
final_prompt = prompt_template.format(language='telugu',query='How are you')
result = llm.invoke(final_prompt).content
print(result)
```

<h2>Chians Runnable</h2>

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model='gpt-5.4')
prompts = [
    {"role":"system","content":"you are a translator and your translating the content to {language}"},
    {"role":"user","content":"{query}"}
]

prompt_template = ChatPromptTemplate.from_messages(prompts)
# runnable = prompt_template.invoke({"language":"Hindi","query":"How are you"})

# result = llm.invoke(runnable).content
# print(result)

chains = prompt_template | llm
result = chains.invoke({"language":"Hindi","query":"How are you"}).content
print(result)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOpenAI(model='gpt-5.4')
output = StrOutputParser()

prompts = [
    {"role":"system","content":"you are a translator and your translating the content to {language}"},
    {"role":"user","content":"{query}"}
]

prompt_template = ChatPromptTemplate.from_messages(prompts)
# runnable = prompt_template.invoke({"language":"Hindi","query":"How are you"})

# result = llm.invoke(runnable).content
# print(result)

chains = prompt_template | llm | output # chain

result = chains.invoke({"language":"Hindi","query":"How are you"})
print(result)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOpenAI(model='gpt-5.4')
output = StrOutputParser()

# custom chain function
def transform_case(result:str):
    return result.upper()

prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"you are a translator and your translating the content to {language}"},
    {"role":"user","content":"{query}"}
])

chains = prompts | llm | output | transform_case

result = chains.invoke({"language":"HinEnglish","query":"i love learning langchain"})
# result = chains.invoke({"language":"English","query":"How are you"})
print(result)
```
