```python
from langchain_groq import ChatGroq

groq_llm = ChatGroq(model="openai/gpt-oss-20b",streaming=True)

text = 'explain genai in detail , i need a complete guideline'
res = groq_llm.stream(text)
print(res)

for chunk in res:
    print(chunk.content, end="")
```
