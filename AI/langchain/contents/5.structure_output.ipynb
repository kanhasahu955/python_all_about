```python
from langchain_openai import ChatOpenAI
from pydantic import BaseModel,Field

llm = ChatOpenAI(model='gpt-5.4')

class ResponseModel(BaseModel):
    name:str = Field(description='name of the user')
    age:int = Field(description="age of the user")
    job:str = Field(description="job of the user")
    company:str = Field(description="user is working on")

structured_output = llm.with_structured_output(ResponseModel)

text = 'i am ashok sahu, and i am 30 years old working '\
'as a software engineer in GSPANN technology'

result = structured_output.invoke(f'form the given text provide me my name, age, job {text}')
print(type(result))

# res = result.model_dump()
print(result.name)
print(result.age)
print(result.job)
print(result.company)


```

```python
from langchain_openai import ChatOpenAI
from pydantic import BaseModel,Field
from typing import List

llm = ChatOpenAI(model='gpt-5.4')

class MovieResponse(BaseModel):
    title:str = Field(description="title of the movie")
    year:int = Field(description="release year")
    genre:str = Field(description="genre of the movie")

class Movies(BaseModel):
    movie:List[MovieResponse]

# structured_output = llm.with_structured_output(MovieResponse)
structured_output = llm.with_structured_output(Movies)

# result = structured_output.invoke('what is the best movie to watch')
result = structured_output.invoke('give me best more scary and dangerious 10 blackmagic  movies in telugu most rated in imdb')

lines = [
    f"{i}. {m.title} ({m.year}) — {m.genre}"
    for i, m in enumerate(result.movie, start=1)
]
print("\n".join(lines))
```
