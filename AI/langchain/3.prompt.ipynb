### prompt templates or dynamic prompting

```python

prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator and you are translating the text from english to {language}"},
    {"role": "user", "content": "{text}"},
])

response = llm.invoke(prompts)
# response = llm.invoke("how to sort the array?")
print(response.content)
```
